```python
adjectives = [
    "Electric", "Silent", "Neon", "Crimson", "Broken",
    "Wild", "Velvet", "Midnight", "Golden", "Savage"
]

nouns = [
    "Wolves", "Echoes", "Rebels", "Dreams", "Storm",
    "Shadows", "Phoenix", "Riders", "Voices", "Flames"
]

funny_words = [
    "Banana", "Pickle", "Potato", "Noodle", "Chicken"
]

metal_words = [
    "Blood", "Skull", "Rage", "Doom", "Inferno"
]
```

```python
import random

def generate_basic_name():
    return f"{random.choice(adjectives)} {random.choice(nouns)}"

def generate_funny_name():
    return f"{random.choice(funny_words)} {random.choice(nouns)}"

def generate_metal_name():
    return f"{random.choice(metal_words)} {random.choice(nouns)}"

def generate_custom_name(city, pet):
    return f"{city} {pet}"

def generate_multiple(n=5):
    return [generate_basic_name() for _ in range(n)]
```

```python
def print_menu():
    print("\n🎸 Band Name Generator")
    print("1. Random Name")
    print("2. Funny Name")
    print("3. Metal Name")
    print("4. Custom Name (City + Pet)")
    print("5. Generate 5 Names")
    print("6. Exit")
```

```python
def main():
    while True:
        print_menu()
        choice = input("Choose an option: ")

        if choice == "1":
            print("👉", generate_basic_name())

        elif choice == "2":
            print("😂", generate_funny_name())

        elif choice == "3":
            print("🔥", generate_metal_name())

        elif choice == "4":
            city = input("Enter your city: ")
            pet = input("Enter your pet name: ")
            print("✨", generate_custom_name(city, pet))

        elif choice == "5":
            names = generate_multiple()
            print("\nGenerated Names:")
            for name in names:
                print("👉", name)

        elif choice == "6":
            print("Goodbye 🎶")
            break

        else:
            print("Invalid choice!")

if __name__ == "__main__":
    main()
```
